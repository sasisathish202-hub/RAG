```python

import os
from dotenv import load_dotenv
from google import genai

# ========================================================================
# 1. LOAD API KEY
# ========================================================================

load_dotenv()

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

# ========================================================================
# 2. AI TEACHER INSTRUCTIONS
# ========================================================================

TEACHER_INSTRUCTIONS = """
You are an expert AI Teacher.

Your job is to teach students, not just provide answers.

Teaching Rules:
- Give only simple explanations using layman terms.
- Answer should be precise and exactly 3 lines.
- Explain the concept simply.
- Use examples whenever necessary.
- Teach students according to their level.
"""

# ========================================================================
# 3. CONVERSATION MEMORY
# ========================================================================

conversation = []

# ========================================================================
# 4. AI TEACHER FUNCTION
# ========================================================================

def ask_teacher(
    question,
    subject="AI & Data science",
    student_level="Beginner",
    language="English"
):

    user_message = f"""
Subject: {subject}

Student Level: {student_level}

Preferred Language: {language}

Student Question: {question}
"""

    # Add user message to conversation history
    conversation.append({
        "role": "user",
        "parts": [{"text": user_message}]
    })

    # Generate Gemini response
    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=conversation,
        config={
            "system_instruction": TEACHER_INSTRUCTIONS
        }
    )

    answer = response.text

    # Store assistant response in conversation history
    conversation.append({
        "role": "model",
        "parts": [{"text": answer}]
    })

    return answer


# ========================================================================
# 5. CHAT APPLICATION
# ========================================================================

def start_teacher():

    print("=" * 60)
    print("                 AI TEACHER")
    print("=" * 60)

    print("Type exit to stop.\n")

    while True:

        question = input("Student: ")

        if question.lower() == "exit":

            print("\nTeacher: Keep learning!")
            break

        answer = ask_teacher(
            question=question,
            subject="AI & Data science",
            student_level="Beginner",
            language="English"
        )

        print("\nTeacher:")
        print(answer)

        print("\n" + "-" * 60)


# ========================================================================
# 6. RUN
# ========================================================================

if __name__ == "__main__":
    start_teacher()
```
